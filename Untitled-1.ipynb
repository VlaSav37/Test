```python
class people:
    def __init__(self,name,age,height,weight):
        self.name = name
        self.age = age
        self.height = height/100
        self.weight = weight 
    def imt(self):
        imt = self.weight/(self.height**2)
        return imt
```

```python
Human = people('Jhon',20,178,75)
Human.imt()
```
